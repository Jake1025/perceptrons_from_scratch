```python
from Neuron import Neuron
from Layer import Layer
from Network import Network
from Activations import Activation

import numpy as np
import random
import pandas as pd
import math

import plotly.express as plt
```

```python
network = Network([
    (3, None), 
    (30, Activation.RELU),
    (30, Activation.SIGMOID),
    (1, Activation.LINEAR)
])
network.initialize_random()
```

```python
training_data = [] # a list of numerical pairs and the answer
testing_data = []
n_examples = 50000

for i in range(n_examples):
    x = random.uniform(0,1)
    y = random.uniform(0,1)
    z = random.uniform(0,1)
    a = x * y * z
    if i%2 == 0:
        training_data.append(((x,y,z), a))
    else:
        testing_data.append(((x,y,z), a))

print(f"Example data point: {training_data[0]}")
```

```python
error = []
for i in range(len(testing_data)):
    error.append(network.forward(testing_data[i][0]) - testing_data[i][1])

print(f"Mean absolute network error: {np.sum(np.abs(error)) / len(testing_data)}")
```

```python
network.train(0.0001, training_data, 10)
```

```python
error = []
for i in range(len(testing_data)):
    error.append(network.forward(testing_data[i][0]) - testing_data[i][1])

print(f"Mean absolute network error: {np.sum(np.abs(error)) / len(testing_data)}")
```

```python
# Compare predictions with x * y * z on the held-out test data.
plot_rows = []
for sample_index, (inputs, target) in enumerate(testing_data):
    x, y, z = inputs
    prediction = float(network.forward(inputs)[0])
    plot_rows.append({
        'sample': sample_index,
        'x': x,
        'y': y,
        'z': z,
        'target': target,
        'prediction': prediction,
        'error': prediction - target,
        'absolute_error': abs(prediction - target),
    })

results = pd.DataFrame(plot_rows)

prediction_plot = plt.scatter(
    results,
    x='target',
    y='prediction',
    color='absolute_error',
    hover_data=['x', 'y', 'z', 'error'],
    title='Test predictions vs. expected product',
    labels={'target': 'Expected x * y * z', 'prediction': 'Network prediction'},
)
min_value = min(results['target'].min(), results['prediction'].min())
max_value = max(results['target'].max(), results['prediction'].max())
prediction_plot.add_shape(
    type='line',
    x0=min_value, y0=min_value, x1=max_value, y1=max_value,
    line={'color': 'black', 'dash': 'dash'},
)
prediction_plot.show()

error_plot = plt.scatter(
    results,
    x='sample',
    y='error',
    color='absolute_error',
    hover_data=['x', 'y', 'z', 'target', 'prediction'],
    title='Test prediction error by sample',
    labels={'sample': 'Test sample', 'error': 'Prediction - target'},
)
error_plot.add_hline(y=0, line_dash='dash', line_color='black')
error_plot.show()
```
