### Build a test Dataset

Were going to train the network to find the normalized sum of two floaing point number in range 0-1. To keep our output in range 0-1 we normalize the sum by dividing by 2.

```python
training_data = [] # a list of numerical pairs and the answer
n_examples = 10000

for i in range(n_examples):
    x = random.uniform(0, 1)
    y = random.uniform(0,1)
    z = x + y
    training_data.append(((x,y),z/2))

print(f"Example data point: {training_data[0]}")
```

### Test the untrained network

First we can use the absolute value of the error for our whole set to determine a baseline for the random network.

```python
error = []
for i in range(len(training_data)):
    error.append(network.forward(training_data[i][0]) - training_data[i][1])

print(f"Total network error: {np.sum(np.abs(error))}")
```
