```python
from Neuron import Neuron
from Layer import Layer
from Network import Network

import numpy as np
import random
import pandas as pd
import math

import plotly.express as plt
```

### Initalize the network randomly

```python
network = Network([2, 1])
network.initialize_random()
```

```python
print(f"Test Inference [0.1, 0.8]: {2 * float(network.forward([0.1, 0.8])[0])}")
```

### Build a test Dataset

Were going to train the network to find the normalized sum of two floaing point number in range 0-1. To keep our output in range 0-1 we normalize the sum by dividing by 2.

```python
training_data = [] # a list of numerical pairs and the answer
n_examples = 50000

for i in range(n_examples):
    x = random.uniform(0, 1)
    y = random.uniform(0,1)
    z = x + y
    training_data.append(((x,y),z/2))

print(f"Example data point: {training_data[0]}")
```

```python
# Compare the trained network with the known (x + y) / 2 relationship
plot_rows = []
for sample_index, sample in enumerate(training_data):
    x, y = sample[0]
    target = sample[1]
    prediction = float(network.forward((x, y))[0])
    plot_rows.append({
        'sample': sample_index,
        'x': x,
        'y': y,
        'target': target,
        'prediction': prediction,
        'error': prediction - target,
        'absolute_error': abs(prediction - target),
    })

results = pd.DataFrame(plot_rows)

prediction_plot = plt.scatter(
    results,
    x='target',
    y='prediction',
    color='absolute_error',
    hover_data=['x', 'y', 'error'],
    title='Predicted output vs. expected output',
    labels={'target': 'Expected (x + y) / 2', 'prediction': 'Network prediction'},
)
min_value = min(results['target'].min(), results['prediction'].min())
max_value = max(results['target'].max(), results['prediction'].max())
prediction_plot.add_shape(
    type='line',
    x0=min_value, y0=min_value, x1=max_value, y1=max_value,
    line={'color': 'black', 'dash': 'dash'},
)
prediction_plot.show()

error_plot = plt.scatter(
    results,
    x='sample',
    y='error',
    color='absolute_error',
    hover_data=['x', 'y', 'target', 'prediction'],
    title='Prediction error by sample',
    labels={'sample': 'Sample', 'error': 'Prediction - target'},
)
error_plot.add_hline(y=0, line_dash='dash', line_color='black')
error_plot.show()
```

### Test the untrained network

First we can use the absolute value of the error for our whole set to determine a baseline for the random network.

```python
error = []
for i in range(len(training_data)):
    error.append(network.forward(training_data[i][0]) - training_data[i][1])

print(f"Absolute network error: {np.sum(np.abs(error))}")
```

### Train the network per example

1) calculate the deltas
2) calculate the gradients

```python
learning_rate = 0.0001
n_runs = 200
bar_width = 30
update_every = max(1, len(training_data) // 100)

for epoch in range(n_runs):
    epoch_loss = 0.0
    print(f"\nEpoch {epoch + 1}/{n_runs}")

    for sample_index, sample in enumerate(training_data):
        network_input = sample[0]
        result = network.forward(network_input)[0]
        error = result - sample[1]
        loss = 0.5 * math.pow(error, 2)
        epoch_loss += loss

        # print(f\"\\n{'=' * 45}\")
        # print(f\"Run {sample_index} | Input: {network_input}\")
        # print(f\"Result: {result:.6f}\")
        # print(f\"Target: {sample[1]:.6f}\")
        # print(f\"Error:  {error:+.6f}\")
        # print(f\"Loss:   {loss:.6f}\")
        # print(f\"{'=' * 45}\")

        if sample_index % update_every == 0 or sample_index == len(training_data) - 1:
            progress = (sample_index + 1) / len(training_data)
            filled = int(bar_width * progress)
            bar = '#' * filled + '-' * (bar_width - filled)
            print(f"\r  [{bar}] {progress:6.1%}", end="", flush=True)


        for l_idx, layer in reversed(list(enumerate(network.layers))):
            for n_idx, neuron in enumerate(layer.neurons):

                derivative = neuron.last_activation * (1 - neuron.last_activation)

                if l_idx == len(network.layers) - 1:
                    delta = error * derivative
                else:
                    next_layer = network.layers[l_idx + 1]

                    propagated_error = sum(
                        next_neuron.delta * next_neuron.weights[n_idx]
                        for next_neuron in next_layer.neurons
                    )

                    delta = propagated_error * derivative

                neuron.delta = delta

                # print(f"  Neuron {n_idx}: activation={neuron.last_activation:.6f}, delta={delta:.6f}")

            

        for layer in network.layers:
            for neuron in layer.neurons:
                weight_gradients = neuron.delta * neuron.last_inputs
                neuron.weights -= learning_rate * weight_gradients
                neuron.bias -= learning_rate * neuron.delta

    average_loss = epoch_loss / len(training_data)
    print(f"\r  [{'#' * bar_width}] 100.0% | average loss: {average_loss:.6f}")
```
