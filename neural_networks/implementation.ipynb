```python
n_weights = 5000

weights = np.random.uniform(-1, 1, n_weights)
inputs = np.random.default_rng().random(n_weights)
bias = random.uniform(0,1)

neuron = Neuron(weights, bias)

neuron.activation(inputs)
```

```python
sample_size = 50000
n_weights = 5000

activations = []
for i in range(sample_size):
    weights = np.random.uniform(-1, 1, n_weights)
    inputs = np.random.default_rng().random(n_weights)
    bias = random.uniform(0,1)
    neuron = Neuron(weights, bias)
    activations.append(neuron.activation(inputs))

df = pd.DataFrame(activations)

np.average(df)
```

```python
fig1 = plt.scatter(
    df,
    template="plotly_dark"
)
fig2 = plt.histogram(
    df,
    template="plotly_dark"
)
fig1.show()
fig2.show()
```
